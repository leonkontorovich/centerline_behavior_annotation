# ome.tif to .avi converter
This script wants to convert the multiple very heavy ome.tif files into one single lighter .avi file


```python
#import pckgs
import cv2
import tifffile as tiff
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
from natsort import natsorted
```

```python
#define compress_ometif function
def compress_ometif(file_path, filename):
    print('reading ome tif file: '+filename+'\n... might take a while')
    retval, mats=cv2.imreadmulti(file_path+filename)
    print('reading complete')
    ometif=np.asarray(mats)
    new_img=np.full((ometif.shape[0],200,200), 0, dtype='uint8')
    print('entering the loop')
    for k,img in enumerate(ometif):
        resized = cv2.resize(img, (200,200), interpolation = cv2.INTER_AREA)
        new_img[k]=resized
    return new_img
    #tiff.imsave(filename,new_img, bigsize=True)
    
```
