```python
# Concatinating a multi file ome.tif into a big tiff ome.btf
# The Concatination is done sequencial (frame by frame)
# The files are closed after concatination
def omeTiffConcatination(sourcePath, storePath):
    # open ome.tiff to be read from
    with tiff.TiffFile(sourcePath) as sourceTiff:
        # store metadata as String
        omexml_string = myTiff.ome_metadata
        # print amount of frames
        print('Frames to be copied: ' + str(len(myTiff.pages.pages)))
        # open file to write
            with tiff.TiffWriter(storePath, bigtiff=True) as storeTiff:
                firstPage = True
                for page in sourceTiff.pages:
                    data = page.asarray()
                    print(page.index)
                if firstPage:
                    storeTiff.save(data, photometric='minisblack',
                    description=omexml_string)
                    firstPage = False
                else:
                    storeTiff.save(data, photometric='minisblack')
```
