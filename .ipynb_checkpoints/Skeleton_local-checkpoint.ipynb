```python
## Tutorial code
from __future__ import print_function
import cv2 as cv
import argparse
parser = argparse.ArgumentParser(description='This program shows how to use background subtraction methods provided by \
                                              OpenCV. You can process both videos and images.')
parser.add_argument('--input', type=str, help='Path to a video or a sequence of image.', default='vtest.avi')
parser.add_argument('--algo', type=str, help='Background subtraction method (KNN, MOG2).', default='MOG2')

args = parser.parse_args()
if args.algo == 'MOG2':
    backSub = cv.createBackgroundSubtractorMOG2()
else:
    backSub = cv.createBackgroundSubtractorKNN()
capture = cv.VideoCapture(cv.samples.findFileOrKeep(args.input))
if not capture.isOpened:
    print('Unable to open: ' + args.input)
    exit(0)


while True:
    ret, frame = capture.read()
    if frame is None:
        break
    
    fgMask = backSub.apply(frame)
    
    
    cv.rectangle(frame, (10, 2), (100,20), (255,255,255), -1)
    cv.putText(frame, str(capture.get(cv.CAP_PROP_POS_FRAMES)), (15, 15),
               cv.FONT_HERSHEY_SIMPLEX, 0.5 , (0,0,0))
    
    
    cv.imshow('Frame', frame)
    cv.imshow('FG Mask', fgMask)
    
    keyboard = cv.waitKey(30)
    if keyboard == 'q' or keyboard == 27:
        break
```

### Adaptive Threshold and Contours approach

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

retval, mats=cv2.imreadmulti('../Data/MMStack.ome.tif')
stack=np.asarray(mats)
img1=stack[0,:,:]
```

```python
worm=img1.copy()
worm=cv2.bitwise_not(worm)

worm_blur=cv2.medianBlur(worm, ksize=9)

#normal threshold
_,worm_binary=cv2.threshold(worm,180,255, cv2.THRESH_BINARY)

#adaptive threshold
#this will be the max value given to px above threshold
max_value=255
#block size from where the adaptive threshold will be calculated
blocksize=9
#value that will be substracted, leave it at 0
C_constant=0
worm_binary=cv2.adaptiveThreshold(worm,max_value,cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, blocksize, C_constant)

#find contours (works on binary)
_, contours, hierarchy= cv2.findContours(worm_binary, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

#draw contours
#-1 stands for all contours
#255 for the color
#1 for the thickness
worm_contours=cv2.drawContours(np.zeros_like(worm), contours,-1,255,1)

plt.figure(figsize=(18,18))
plt.subplot(221)
plt.imshow(worm, cmap='gray')
plt.title('original')
plt.subplot(222)
plt.imshow(worm_blur, cmap='gray')
plt.title('blur')
plt.subplot(223)
plt.imshow(worm_binary, cmap='gray')
plt.title('binary')
plt.subplot(224)
plt.imshow(worm_contours, cmap='gray')
plt.title('con')
```

#### Trying different thresholding methods and Floodfill

```python
#Threshold + Floodfill
#Example of standard threshold + adaptative thresholds using two different blocksize.
#blocksize small gives you more detailed features (more precise nose position),
#but big blocksize is more robust

#Flodfill can work in two ways, flooding the background or flooding the inside.

input_image=sub.copy()

#normal thresholding
ret, worm_binary0 = cv2.threshold(input_image,19,255,cv2.THRESH_BINARY)
h, w = input_image.shape[:2]
#the mask is 2 px higher and wider
mask0 = np.zeros((h+2, w+2), np.uint8)
#to flood from inside
background_pixel0=(305,306)
#to flood from outside
#background_pixel0=(494,45)
cv2.floodFill(worm_binary0.copy(), mask0, background_pixel0, 255);

#this will be the max value given to px above threshold
max_value=255
#block size from where the adaptive threshold will be calculated
#it is robust to high values
blocksize=11
#value that will be substracted, leave it at 0
C_constant=0
worm_binary=cv2.adaptiveThreshold(input_image,max_value,cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, blocksize, C_constant)

blocksize2=51
worm_binary2=cv2.adaptiveThreshold(input_image,max_value,cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, blocksize2, C_constant)

#img floodfill
img_floodfill=worm_binary.copy()
print(type(img_floodfill))
h, w = img_floodfill.shape[:2]
#the mask is 2 px higher and wider
mask = np.zeros((h+2, w+2), np.uint8)
background_pixel=(494,45)
#im flodfill contains the original image, mask where it will be applied

if worm_binary[tuple(background_pixel)]==0:
    cv2.floodFill(img_floodfill, mask, background_pixel, 255);
else:
    print('Choose antoher background pixel')

#second image
mask2 = np.zeros((h+2, w+2), np.uint8)    
img_floodfill2=worm_binary2.copy()

background_pixel2=(494,45)
#im flodfill contains the original image, mask where it will be applied

if worm_binary2[tuple(background_pixel)]==0:
    cv2.floodFill(img_floodfill2, mask2, background_pixel2, 255);
else:
    print('Choose antoher background pixel')

#plotting
rows, cols=3, 2
plt.figure(figsize=(18,18+9))
plt.subplot(rows, cols,1)
plt.imshow(worm_binary0, cmap='gray')
plt.title('worm binary normal threshold')
plt.subplot(rows, cols,2)
plt.imshow(mask0, cmap='gray')
plt.title('worm binary MASK normal threshold')
plt.subplot(rows, cols,3)
plt.imshow(worm_binary, cmap='gray')
plt.title('worm binary blocksize '+ str(blocksize))
plt.subplot(rows, cols,4)
plt.imshow(mask, cmap='gray')
plt.title('worm binary MASK blocksize '+ str(blocksize))
plt.subplot(rows, cols,5)
plt.imshow(worm_binary2, cmap='gray')
plt.title('worm binary blocksize '+ str(blocksize2))
plt.subplot(rows, cols,6)
plt.imshow(mask2, cmap='gray')
plt.title('worm binary MASK blocksize '+ str(blocksize2))

cv2.imwrite('worm_binary2.tiff', worm_binary2)
cv2.imwrite('mask2.tiff', mask2)
cv2.imwrite('mask0.tiff', mask0)
```

#### Actual preprocessing
Cleaning image to have only the worm

Idea is to perform two floodfill operations to get rid of the particles outside the worm, and have the worm filled.

To detect a pixel in the center of the worm I take a ROI in the center, and find the max of the distance transformation. This should be inside the worm

```python
#PREPROCESSING for SCIKIT approach: Centroid
img=sub.copy()

img_blur=cv2.blur(img,(10,10))

M=cv2.moments(img_blur)

# calculate x,y coordinate of centroid
cX = int(M["m10"] / M["m00"])
cY = int(M["m01"] / M["m00"])
#draw centroid
cv2.circle(img, (cX, cY), 5, (255, 0, 0), -1)

plt.figure(figsize=(5,5))
plt.imshow(img, cmap='gray')
plt.title('Centroid falls outside the worm')
```

```python
#the approach now is to take a roi in the center, calculate MAX of the distance transformation
#and take that coordinate to do an inverse floodfill.
img=sub.copy()

roi_len=200
ch, cw= np.asarray(img.shape, dtype=int)//2
print('center of the image: ' +str(ch) +'(height)'' '+  str(cw)+'(width)')

roi=img[ch-roi_len//2:ch+roi_len//2, cw-roi_len//2:cw+roi_len//2]

dist = cv2.distanceTransform(roi, cv2.DIST_L2, 3)

#coordinates of the first max coordinate
center_x=np.where(dist==dist.max())[1][0]
center_y=np.where(dist==dist.max())[0][0]
print(np.where(dist==dist.max()))
center_px=(center_x,center_y)

roi_copy=roi.copy()
cv2.circle(roi_copy, center=(center_px[0], center_px[1]), radius=10, color=(255,255,255), thickness=2)
print('coordinate inside the worm roi: ' +str(center_px))


#calculate the center relative to the whole image
cx=(ch-roi_len//2)+center_px[0]
cy=(cw-roi_len//2)+center_px[1]
center=(cx,cy)
center
print('coordinate in the entire image: ' +str(center))

#plotting
r,c=2,3

plt.figure(figsize=(18,9))
plt.subplot(r,c,1)
plt.imshow(img)
plt.title('complete img')
plt.subplot(r,c,2)
pt1=(ch-roi_len//2,cw-roi_len//2)
pt2=(ch+roi_len//2,cw+roi_len//2)
img_rect=img.copy()
cv2.rectangle(img_rect, pt1,pt2,(255,255,255), 10)
plt.imshow(img_rect)
plt.title('complete img with ROI')
plt.subplot(r,c,3)
plt.imshow(roi)
plt.title('ROI')
plt.subplot(r,c,4)
plt.imshow(dist)
plt.title('dist. transf.')
plt.subplot(r,c,5)
plt.imshow(roi_copy)
plt.title('MAX of dist. transf.')
plt.subplot(r,c,6)
img_circ=img_rect.copy()
cv2.circle(img_circ, center=(center[0], center[1]), radius=10, color=(255,0,0), thickness=2)
plt.imshow(img_circ)
plt.title('complete img with roi and circle')
plt.show()
cv2.imwrite('dist.tiff',dist)
```

#### Scikit graph approach
I took it from the second answer from here:
    
    https://stackoverflow.com/questions/51043824/how-to-find-shortest-path-in-skeletonized-maze-image
    
It works! Somehow at the beginning I got an error and I thought it does not work on binary images.

Now it works and it gives me the ordered pixels.


https://scikit-image.org/docs/dev/api/skimage.graph.html#route-through-array


```python
#Scikit graph approach

import skimage.graph

binary=skeleton.copy()

### give start (y1,x1) and end (y2,x2) and the binary maze image as input
def shortest_path(start,end,binary):
    costs=np.where(binary,1,1000)
    path, cost = skimage.graph.route_through_array(costs, start=start, end=end, fully_connected=False)
    return path,cost,costs

#manually entered
start_point =  (480, 332)#(332,480)#
end_point = (124, 286)#(286, 124)#

corners=corners3.copy()

#somehow the coordinates have to be entered the otherway around [0,0,1] first, [0,0,0] last
start_point=corners[0,0,1],corners[0,0,0]
end_point=corners[1,0,1],corners[1,0,0]

#from nose/tail detection

path, cost, costs=shortest_path(start_point, end_point, binary)

#somehow the coordinates are the other way around
y, x = zip(*path)
fig=plt.figure(figsize=(15,10), dpi=100)
plt.subplot(121)
plt.imshow(binary)
plt.subplot(121)
#fig=plt.figure(figsize=(610,608), dpi=100)

plt.plot(x,y, 'o', linestyle='dashed', linewidth=.2, markersize=1)
plt.savefig('connected.tiff')
# fig2=plt.figure(figsize=(15,10), dpi=100)
# plt.imshow(costs[123:140,285:310])
```

```python
#find spline with interp1d
from scipy.interpolate import interp1d

#trying iverted npa, IT WORKS!
new_npa=npa.copy()
new_npa[:,0]=npa[:,1]
new_npa[:,1]=npa[:,0]

points=new_npa #npa

# Linear length along the line:
distance = np.cumsum( np.sqrt(np.sum( np.diff(points, axis=0)**2, axis=1 )) )
distance = np.insert(distance, 0, 0)/distance[-1]

# Interpolation for different methods:
interpolations_methods = ['slinear', 'quadratic', 'cubic']
alpha = np.linspace(0, 1, 75)

interpolated_points = {}
for method in interpolations_methods:
    interpolator =  interp1d(distance, points, kind=method, axis=0)
    interpolated_points[method] = interpolator(alpha)

# Graph:
plt.figure(figsize=(5,5))
for method_name, curve in interpolated_points.items():
    #plt.figure()
    plt.plot(*curve.T, '-', label=method_name);
    plt.axis('equal')
    
#plt.figure()
plt.plot(*points.T, 'ok', label='original points');
plt.axis('equal'); plt.legend(); 
#plt.xlabel('x'); plt.ylabel('y');
```

```python
#find spline with interp1d
from scipy.interpolate import interp1d

pts=npa
distance=np.cumsum(np.sqrt(np.sum(np.diff(pts, axis=0)**2, axis=1)))
distance = np.insert(distance, 0, 0)/distance[-1]

# Interpolation for different methods:
interpolations_methods = ['slinear', 'quadratic', 'cubic']
alpha = np.linspace(0, 1, 75)

interpolator_linear =  interp1d(distance, pts, kind='slinear', axis=0)
interpolator_quadratic =  interp1d(distance, pts, kind='quadratic', axis=0)
interpolator_cubic =  interp1d(distance, pts, kind='cubic', axis=0)

interpolated_linear_pts = interpolator_linear(alpha)
interpolated_quadratic_pts = interpolator_linear(alpha)
interpolated_cubic_pts = interpolator_linear(alpha)


plt.figure(figsize=(15,2))
plt.subplot(131)
plt.plot(*interpolated_linear_pts.T, '-', label='linear');
plt.title('linear')
plt.subplot(132)
plt.plot(*interpolated_quadratic_pts.T, '-', label='quadratic');
plt.title('quadratic')
plt.subplot(133)
plt.plot(*interpolated_cubic_pts.T, '-', label='cubic');
plt.title('cubic')

```

```python
#define shortest_path function
def shortest_path(start,end,binary):
    costs=np.where(binary,1,1000)
    path, cost = skimage.graph.route_through_array(costs, start=start, end=end, fully_connected=False)
    return path,cost,costs
```

```python
#it seems what I was doing with floodfill is pretty much BS.
#I should use findContours, connectedComponentWithStats and Moments.

n_frames=5000
new_video=video[0:n_frames].copy()
#new_video=video.copy()

# head_x=head_x[4500:-1].copy()
# head_y=head_y[4500:-1].copy()
# tail_x=tail_x[4500:-1].copy()
# tail_y=tail_y[4500:-1].copy()

# n_frames=100
# new_video=video[0:n_frames].copy()
#os.remove('second_der.csv')

sec_der=[]
knots=100

all_x_new=np.zeros((new_video.shape[0],knots))
all_y_new=np.zeros((new_video.shape[0],knots))
all_der=np.zeros((new_video.shape[0],knots))
all_K=np.zeros((new_video.shape[0],knots))


#new_video=cv2.bitwise_not(new_video)
#tiff.imsave('input_tiff.tiff',new_video)

#for every frame (img) k
for k, img in enumerate(new_video):

    img[:]=cv2.bitwise_not(img)
    
    #substrack background and store it in sub
    img[:]=cv2.subtract(img,bg_img)
    
    #median Blur
    img[:] = cv2.medianBlur(img,5)
    #apply threshold
    ret, new_img = cv2.threshold(img,11,255,cv2.THRESH_BINARY)
    #find contours
    contours, hierarchy = cv2.findContours(new_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    #list areas of contours, find MAX, draw contours from MAX area
    areas=[]
    for i in range(0, len(contours)):
        areas.append(cv2.contourArea(contours[i]))
    
    worm_contour=np.where(areas==np.asarray(areas).max())
    worm_contour=np.asarray(worm_contour)
    
    img_contours = np.zeros(img.shape)
    img[:]=cv2.drawContours(img_contours,contours, worm_contour, 255, -1)
    #cv2.circle(img, (int(head_x[k]), int(head_y[k])), 5, (255,255,255), 2)
    #cv2.circle(img, (int(tail_x[k]), int(tail_y[k])), 5, (255,255,255), 2)
    
    #centerline part:
    #run skeletonize
    img_copy=img/255
    skeleton = skeletonize(img_copy)
    
    #Scikit graph approach

    ### give start (y1,x1) and end (y2,x2) and the binary maze image as input
    #shortest path

    start_point =  (int(head_y[k]), int(head_x[k]))#(286, 124)#
    end_point = (int(tail_y[k]), int(tail_x[k]))#(332,480)#

    path, cost, costs=shortest_path(start_point, end_point, skeleton)
    
    #somehow the coordinates are the other way around
    y, x = zip(*path)
#     plt.figure()
#     plt.imshow(img)
#     plt.plot(x,y)
    
    #
    pts=np.asarray(path, dtype=np.int)

    ####
    #s is the smoothing condition should have around the size of points/2 (keep it low)
    #k is the degree of freedom for the polynom it fits, 5 is good
    tck, u = splprep(pts.T, u=None, s=pts.shape[0]/2, per=0, k=5)
    u_new = np.linspace(u.min(), u.max(), knots)#1000)

    x_new, y_new = splev(u_new, tck, der=0)

    #this returns x'(s), y'(s)
    x_der, y_der = splev(u_new, tck, der=1)
    #to have y'(x):
    der=y_der/x_der

    #this returns x''(s), y''(s)
    x_der2, y_der2 = splev(u_new, tck, der=2)
    #to have y''(x), also called K for Curvature:
    #we need the following equation:
    #ref in: https://en.wikipedia.org/wiki/Curvature#In_terms_of_a_general_parametrization (1st equation)
    K=(x_der*y_der2-y_der*x_der2)/np.sqrt(x_der**2+y_der**2)**3
    ####
    
    #np.savetxt("second_der.csv", y_der2, delimiter=",")

    for point in path:
        y, x = zip(point)
        img[y,x]=1

    
    #append in a variable
    #all_pts[k]=pts
    all_x_new[k]=x_new
    all_y_new[k]=y_new
    all_der[k]=der
    all_K[k]=K
    
    #append_list_as_row('first_der.csv', y_der)
    #append_list_as_row('second_der2.csv', K)
    
    img[:]=img

output_path='/Users/ulises.rey/code/python/skeleton/'    
tiff.imsave(output_path+'output_bigtiff2.tiff',new_video, bigtiff=True)
np.savetxt(output_path+'second_der_np_new.csv', all_K, delimiter=",")
np.savetxt(output_path+'x_new_np.csv', all_x_new, delimiter=",")
np.savetxt(output_path+'y_new_np.csv', all_y_new, delimiter=",")
#tiff.imsave('new_img.tiff',new_img)
```
